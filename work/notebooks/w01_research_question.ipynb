## 4. Careful words — what I can and can't claim

### ✅ What I can claim (directional):
- Pages with low CTR, high average position, and low query diversity tend to be at higher risk of declining.
- A Random Forest model can rank pages by decline risk with better precision than a basic hand-written rule.
- These patterns are **observed** in the anonymized dataset provided by FlyRank.

### ❌ What I cannot claim (causal):
- I cannot claim that updating a page **will** reverse its decline (correlation ≠ causation).
- I cannot claim this model will work on every client without re-training (needs client-holdout validation).
- I cannot claim to "predict Google's algorithm" — I am only predicting traffic trends.

### ⚠️ Why this matters:
The goal is to help humans prioritize their work, not to automate the decision entirely. The model suggests *where* to look; the human decides *what* to do.
